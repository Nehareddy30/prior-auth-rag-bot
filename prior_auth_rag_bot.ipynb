```python
from fpdf import FPDF

class PolicyPDF(FPDF):
    def header(self):
        self.set_font("Helvetica", "B", 11)
        self.cell(0, 10, "ACME Health Plan - Coverage Policy Document", align="C", new_x="LMARGIN", new_y="NEXT")
        self.set_font("Helvetica", "", 9)
        self.cell(0, 6, "Confidential - For Internal Use Only", align="C", new_x="LMARGIN", new_y="NEXT")
        self.ln(4)

    def footer(self):
        self.set_y(-15)
        self.set_font("Helvetica", "I", 8)
        self.cell(0, 10, f"Page {self.page_no()}", align="C")

def make_pdf(title, policy_num, sections, outpath):
    pdf = PolicyPDF()
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()
    pdf.set_font("Helvetica", "B", 14)
    pdf.multi_cell(0, 10, title, new_x="LMARGIN", new_y="NEXT")
    pdf.set_font("Helvetica", "", 10)
    pdf.multi_cell(0, 6, policy_num)
    pdf.ln(2)
    for sec_title, body in sections:
        pdf.set_font("Helvetica", "B", 11)
        pdf.multi_cell(0, 7, sec_title, new_x="LMARGIN", new_y="NEXT")
        pdf.set_font("Helvetica", "", 10)
        pdf.multi_cell(0, 6, body, new_x="LMARGIN", new_y="NEXT")
        pdf.ln(3)
    pdf.output(outpath)
    print(f"✓ Created: {outpath}")

make_pdf(
    "Musculoskeletal Imaging Authorization Policy",
    "Policy Number: MSK-IMG-2024-001 | Effective: January 1, 2024",
    [
        ("1. PURPOSE",
         "This policy establishes prior authorization requirements for musculoskeletal imaging services, including MRI, CT, and X-ray studies of the spine, joints, and extremities."),
        ("2. DEFINITIONS",
         "Conservative Treatment: Non-surgical treatment including physical therapy, chiropractic manipulation, and pharmacologic management for a minimum of six (6) weeks.\n\nRed Flag Symptoms: Clinical findings warranting immediate imaging without prior conservative treatment, including cauda equina syndrome, severe neurological deficit, fracture, malignancy, or infection."),
        ("3.1 Low Back Pain - Lumbar Spine MRI (CPT 72148, 72149, 72158)",
         "Authorization approved when ALL criteria met:\n\n"
         "a) Confirmed diagnosis of low back pain (ICD-10: M54.5).\n"
         "b) Minimum six (6) weeks of conservative treatment completed.\n"
         "c) Provider notes with dates of service documented.\n\n"
         "EXCEPTION: Waived for members with red flag symptoms."),
        ("3.2 Cervical Spine MRI (CPT 72141, 72142, 72156)",
         "Authorization approved when:\n\n"
         "a) Cervical radiculopathy (ICD-10: M54.12) or myelopathy (ICD-10: M47.812) diagnosed.\n"
         "b) Four (4) weeks conservative treatment completed.\n"
         "c) Neurological examination findings documented."),
        ("3.3 Knee MRI (CPT 73721, 73722)",
         "Authorization approved when:\n\n"
         "a) Ligament tear (ICD-10: S83), meniscus tear (ICD-10: M23.2), or osteochondral defect diagnosed.\n"
         "b) X-ray performed within prior 12 months.\n"
         "c) Four (4) weeks of physical therapy completed."),
        ("4. DOCUMENTATION REQUIREMENTS",
         "Providers must submit:\n"
         "- PA request form with member ID and date of birth\n"
         "- Clinical notes with diagnosis and treatment history\n"
         "- Conservative treatment records with dates\n"
         "- Ordering physician NPI number\n\n"
         "Incomplete requests returned within two (2) business days."),
        ("5. AUTHORIZATION TIMEFRAMES",
         "Standard requests: three (3) business days.\n"
         "Expedited requests: one (1) business day when delay jeopardizes member health."),
    ],
    "data/sample_policies/musculoskeletal_imaging_policy.pdf"
)

make_pdf(
    "Behavioral Health and Telehealth Services Policy",
    "Policy Number: BH-TEL-2024-003 | Effective: January 1, 2024",
    [
        ("1. PURPOSE",
         "This policy defines prior authorization requirements for behavioral health services and telehealth visits for ACME Health Plan members."),
        ("2. TELEHEALTH COVERAGE - BEHAVIORAL HEALTH",
         "Telehealth visits covered when:\n\n"
         "a) Services by licensed behavioral health provider credentialed with ACME Health Plan.\n"
         "b) HIPAA-compliant video platform used.\n"
         "c) Services clinically appropriate for telehealth.\n\n"
         "CPT codes: 90791, 90834, 90837, 90847, 90853 (modifier 95 or GT)."),
        ("3. INPATIENT BEHAVIORAL HEALTH",
         "Inpatient psychiatric admissions (ICD-10: F20-F99) require prior authorization except emergencies.\n\n"
         "Criteria:\n"
         "- Active suicidal or homicidal ideation with plan and intent\n"
         "- Psychotic break or acute manic episode\n"
         "- Substance detoxification requiring 24-hour monitoring\n\n"
         "Reviewed every five (5) days for continued stay."),
        ("4. BARIATRIC SURGERY AUTHORIZATION",
         "Bariatric surgery (CPT 43644, 43645, 43770) requires ALL criteria:\n\n"
         "a) BMI >= 40, OR BMI >= 35 with comorbidity (diabetes ICD-10 E11, hypertension I10, sleep apnea G47.33).\n"
         "b) Six (6) months supervised weight loss program documented.\n"
         "c) Psychological evaluation within 12 months prior to surgery.\n"
         "d) Cardiologist clearance if cardiovascular disease present.\n"
         "e) No active substance use disorder or uncontrolled psychiatric condition."),
        ("5. SUBSTANCE USE DISORDER",
         "Covered without prior authorization for first 14 days:\n"
         "- Outpatient SUD counseling (CPT 99408, 99409)\n"
         "- Medication-Assisted Treatment: buprenorphine, naltrexone (ICD-10: F11.20)\n\n"
         "Residential SUD treatment (HCPCS H0018) requires prior authorization."),
    ],
    "data/sample_policies/behavioral_health_telehealth_policy.pdf"
)

print("\nAll policy PDFs created successfully!")
```

```python
from pypdf import PdfReader

def load_pdfs(policies_dir):
    all_docs = []
    pdf_files = [f for f in os.listdir(policies_dir) if f.endswith(".pdf")]
    
    print(f"Found {len(pdf_files)} policy PDF(s):\n")
    for filename in pdf_files:
        filepath = os.path.join(policies_dir, filename)
        reader = PdfReader(filepath)
        for page_num, page in enumerate(reader.pages):
            text = page.extract_text()
            all_docs.append({
                "source_file": filename,
                "page": page_num + 1,
                "content": text
            })
        print(f"✓ {filename} — {len(reader.pages)} page(s) loaded")
    
    print(f"\nTotal pages loaded: {len(all_docs)}")
    return all_docs

docs = load_pdfs("data/sample_policies")

# Preview first document
print("\n--- Preview of first page ---")
print(f"Source: {docs[0]['source_file']}")
print(f"Page: {docs[0]['page']}")
print(f"\nContent:\n{docs[0]['content'][:400]}")
```

```python
def split_into_chunks(docs, chunk_size=800, overlap=150):
    chunks = []
    
    for doc in docs:
        text = doc["content"]
        start = 0
        
        while start < len(text):
            end = start + chunk_size
            chunk_text = text[start:end]
            
            chunks.append({
                "source_file": doc["source_file"],
                "page": doc["page"],
                "content": chunk_text
            })
            
            # Move forward but overlap by 150 chars for context continuity
            start += chunk_size - overlap
    
    return chunks

chunks = split_into_chunks(docs)

print(f"Total chunks created: {len(chunks)}")
print(f"\n--- Sample chunk ---")
print(f"Source: {chunks[0]['source_file']}")
print(f"Page: {chunks[0]['page']}")
print(f"Length: {len(chunks[0]['content'])} characters")
print(f"\nContent preview:\n{chunks[0]['content'][:300]}")
```

```python
import numpy as np
import faiss
from sklearn.feature_extraction.text import TfidfVectorizer

# We'll use TF-IDF vectors (no API key needed, works offline)
# TF-IDF finds chunks based on important words in your question

# Extract just the text from chunks
texts = [chunk["content"] for chunk in chunks]

# Build TF-IDF vectors
print("Building search index...")
vectorizer = TfidfVectorizer(
    stop_words="english",    # ignore common words like "the", "is"
    ngram_range=(1, 2),      # match single words AND two-word phrases
    max_features=5000
)

# Fit and transform all chunk texts into vectors
vectors = vectorizer.fit_transform(texts).toarray().astype("float32")

# Normalize vectors so similarity scores are consistent
faiss.normalize_L2(vectors)

# Build FAISS index
dimension = vectors.shape[1]
index = faiss.IndexFlatIP(dimension)   # IP = Inner Product (cosine similarity)
index.add(vectors)

print(f"✓ Index built with {index.ntotal} chunks")
print(f"✓ Each chunk represented as {dimension} dimensional vector")
print("\nSearch index is ready!")
```

```python
def retrieve_chunks(question, k=3):
    """
    Given a question, find the top-k most relevant policy chunks.
    """
    # Convert question into the same vector format as our chunks
    question_vector = vectorizer.transform([question]).toarray().astype("float32")
    faiss.normalize_L2(question_vector)
    
    # Search the index for the most similar chunks
    scores, indices = index.search(question_vector, k)
    
    results = []
    for score, idx in zip(scores[0], indices[0]):
        results.append({
            "source_file": chunks[idx]["source_file"],
            "page": chunks[idx]["page"],
            "content": chunks[idx]["content"],
            "similarity_score": round(float(score), 4)
        })
    
    return results

# Test it with a sample question
question = "Is an MRI of the lumbar spine covered without prior conservative treatment?"

print(f"Question: {question}")
print("\n--- Top 3 Relevant Policy Sections ---\n")

results = retrieve_chunks(question, k=3)

for i, result in enumerate(results, 1):
    print(f"[{i}] Source: {result['source_file']} | Page: {result['page']} | Score: {result['similarity_score']}")
    print(f"    Preview: {result['content'][:200]}")
    print()
```

```python
def generate_answer(question, results):
    """
    Takes the retrieved policy chunks and formats a clear cited answer.
    """
    print(f"QUESTION:\n{question}")
    print("\n" + "="*60)
    print("RELEVANT POLICY SECTIONS FOUND:")
    print("="*60)
    
    # Show the most relevant chunks with citations
    for i, result in enumerate(results, 1):
        print(f"\n[Source {i}] {result['source_file']} — Page {result['page']}")
        print(f"Relevance Score: {result['similarity_score']}")
        print("-"*40)
        print(result['content'])
    
    print("\n" + "="*60)
    print("ANSWER BASED ON POLICY DOCUMENTS:")
    print("="*60)
    
    # Use the highest scoring chunk as the primary answer source
    top_chunk = results[0]
    second_chunk = results[1] if len(results) > 1 else None
    
    print(f"\nBased on: {top_chunk['source_file']} (Page {top_chunk['page']})")
    print()
    
    # Extract and display the most relevant content
    combined_content = top_chunk['content']
    if second_chunk and second_chunk['similarity_score'] > 0.15:
        combined_content += "\n\n" + second_chunk['content']
    
    print(combined_content)
    print("\n" + "="*60)
    print("NOTE: Always verify with the full policy document.")
    print("="*60)

# Test with our question
results = retrieve_chunks(question, k=3)
generate_answer(question, results)
```

```python
import json
from datetime import datetime

def log_query(question, results):
    """
    Saves each query to a log file — important for audit trails
    in HIPAA-regulated healthcare environments.
    """
    log_file = "outputs/query_log.jsonl"
    
    log_entry = {
        "timestamp": datetime.utcnow().isoformat(),
        "question": question,
        "sources_retrieved": [
            {
                "source_file": r["source_file"],
                "page": r["page"],
                "similarity_score": r["similarity_score"]
            }
            for r in results
        ]
    }
    
    with open(log_file, "a") as f:
        f.write(json.dumps(log_entry) + "\n")
    
    print(f"✓ Query logged to {log_file}")

# Test it
log_query(question, results)

# Show the log file
print("\n--- Query Log ---")
with open("outputs/query_log.jsonl", "r") as f:
    print(f.read())
```

```python
def ask(question):
    """
    Full pipeline: question → retrieve → log → display answer.
    Call this function with any prior auth question.
    """
    print("\n" + "="*60)
    print(f"QUESTION: {question}")
    print("="*60)
    
    # Retrieve relevant chunks
    results = retrieve_chunks(question, k=3)
    
    # Log the query
    log_query(question, results)
    
    # Display cited answer
    print("\nMOST RELEVANT POLICY SECTIONS:")
    print("-"*60)
    for i, result in enumerate(results[:2], 1):
        print(f"\n[Source {i}] {result['source_file']} — Page {result['page']}")
        print(f"Score: {result['similarity_score']}")
        print(result['content'][:600])
    
    print("\n" + "="*60)
    print("NOTE: Always verify with the full policy document.")
    print("="*60)

# Try these questions one by one!
ask("What documentation is required for bariatric surgery authorization?")
```
